### Remove columns with more than 10% missing values

```python
# https://thispointer.com/pandas-drop-dataframe-columns-based-on-nan-percentage/
perc = 10.0
min_count =  int(((100-perc)/100)*data.shape[0] + 1)
mod_df = data.dropna( axis=1, 
                thresh=min_count)
print("Modified Dataframe : ")
print(mod_df)
```

```python
mod_df.shape 
```

### Remove rows with more than 20% missing values;

```python
perc = 25.0 # Here N is 75
min_count =  int(((100-perc)/100)*data.shape[1] + 1)
mod_df = data.dropna( axis=0, 
                    thresh=min_count)
print(mod_df)
```
