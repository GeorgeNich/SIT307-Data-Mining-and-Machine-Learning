**Q1:** Create a list of randomly selected 200 integers between 0 and 1000 and print in ascending order all the odd numbers that are divisible by 7.

```python
import random

randomlist = random.sample(range(0, 1001), 200)
divsev_list = []

for div7 in randomlist:
    if div7 % 2 != 0 and div7 % 7 == 0:
        divsev_list.append(div7)

divsev_list.sort()
print(divsev_list)
```

**Q2:** Write a python code to calculate Body Mass Index (the ratio of weight in kilograms (kg) divided by height in metres-squared (m2)). Write the initial input commands to enter "weight" and "height", round Body Mass Index to 1 decimal place and print.

```python
heightinput = float(input("Enter precise weight in (cm): "))
weightinput = float(input("Enter precise height in (kg): "))

BMI = weightinput / (heightinput/100)**2


print(str(round(BMI, 1)) + " Is the calculated BMI for given weight and height")
```

**Q3:** Equilateral triangle all sides are equal. Scalene triangle none of the sides are equal. Isosceles triangle atleast two sides are equal.
Set the vertices A, B and C as below. 

i): Write a python program to calculate and print the sides (use 2 decimal) a = |CB|, b = |AC| and c = |BC| and check if the triangle is equilateral/isosceles/scalene. 

ii): Calculate and print the area of the triangle by the following formula: 
$s = (a+b+c)/2;$  
Area = $\sqrt(s(s-a)*(s-b)*(s-c))$.

```python

#reference code

import numpy as np
import math

A = np.random.randint(low=2, high=100, size=2)
B = np.random.randint(low=2, high=100, size=2)
C = np.random.randint(low=2, high=100, size=2)
print(A,B,C)

a = np.linalg.norm(B - C)
b = np.linalg.norm(A - C)
c = np.linalg.norm(A - B)

roundingup_a = str(round(a,2))
roundingup_b = str(round(b,2))
roundingup_c = str(round(c,2))
print(roundingup_a,roundingup_b,roundingup_c)

if roundingup_a == roundingup_b == roundingup_c:
    print("This is a equilateral Triangle") #equalateral very rare due to random
    
    #2 sides are the same = isoceles
elif roundingup_a == roundingup_b or roundingup_b == roundingup_c or roundingup_c == roundingup_a:
    print("This is a isosceles Traiangle")
else:
    print("This is a scalene Traiangle") #else if scalene

S = (a+b+c)/2    
Area = math.sqrt(S*(S-a)*(S-b)*(S-c)) 

print("A's length: " + str(round(a, 2)) + " B's length: " + str(round(b,2)) +" C's length: " + str(round(c,2)))
print("Area: " + str(round(Area, 1)))
print("S equals:" + str(round(S, 2)))

#https://www.geeksforgeeks.org/calculate-the-euclidean-distance-using-numpy/
```

**Q4:** In a Dataframe create an array in range(10-3000) in one column and a true or false column adjacent to this column which would be true if the number is a palindrome else false. Also print the total number of true values in this column.

```python
#https://www.tutorialspoint.com/python-program-to-check-if-a-given-string-is-number-palindrome
#https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.append.html
#reference code for checking pailndrome

import pandas as pd
import math

#create title columns
df = pd.DataFrame(columns = ['Number', 'isitaPalindrome'])

#variable to hold count of True
Totalpal = 0
#create range of numbers
randomarr = range(10, 3001)


#loop through list and check if number is palindrome
for i in randomarr:
    if i == int(str(i)[::-1]): #if the number is the same going backwards as the initial starting number
        df = df.append({'Number': i, 'isitaPalindrome': True}, ignore_index = True) #append number and True
        Totalpal += 1
    else:
        df = df.append({'Number': i, 'isitaPalindrome': False}, ignore_index = True) # append number and false

#print all data
#pd.set_option('display.max_rows', None, 'display.max_columns', None)
print(df)
print("Total true: ", Totalpal)
```
