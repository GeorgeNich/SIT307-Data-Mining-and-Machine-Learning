```python
import matplotlib.pyplot as plt
import pandas as pd
data = pd.read_csv("data/iris.txt")
```

```python
data.describe()

# used data.describe() to give an overview of the data we are using
```

```python
x = data[ " petal length"]
y = data[ " petal width"]

plt.boxplot(x)
plt.title("petal length")
plt.show()

plt.boxplot(y)
plt.title("petal width")
plt.show()
```

#### used a boxplot to portray and compare the data from both petal length and petal width for an understanding into the quartiles of said data being displayed against one another as opposed to being included in the same data graph, i felt box plots were a good way at getting insight to the x and y variables seperatelt as opposed to what the numbers would portray when used in conjunction with eachother in the example scatter graph and line graph
